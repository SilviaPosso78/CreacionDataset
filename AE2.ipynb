# Actividad 2: Manejo de archivos CSV, JSON y TXT

Este notebook importa, modifica y guarda un archivo CSV y un archivo JSON descargados manualmente de Hugging Face, usando una clase para cada formato. Al final, una tercera clase genera un resumen en `resumen.txt`.

## 1. Archivos de origen (descarga manual desde Hugging Face)

- **CSV:** `Iris.csv` — https://huggingface.co/datasets/scikit-learn/iris/blob/main/Iris.csv
- **JSON:** `code_alpaca_20k.json` — https://huggingface.co/datasets/sahil2801/CodeAlpaca-20k/blob/main/code_alpaca_20k.json

Ambos archivos se descargaron manualmente y se ubicaron en la misma carpeta de este notebook. Los originales no se modifican: los cambios se guardan en archivos nuevos.

```python
# URL específicas de origen (descarga manual desde Hugging Face)
url_csv_origen = "https://huggingface.co/datasets/scikit-learn/iris/blob/main/Iris.csv"
url_json_origen = "https://huggingface.co/datasets/sahil2801/CodeAlpaca-20k/blob/main/code_alpaca_20k.json"

# Rutas relativas de los archivos locales
archivo_csv_original = "Iris.csv"
archivo_json_original = "code_alpaca_20k.json"
archivo_csv_modificado = "Iris_modificado.csv"
archivo_json_modificado = "code_alpaca_20k_modificado.json"

print("URL CSV: ", url_csv_origen)
print("URL JSON:", url_json_origen)
```

## 2. Definición de las clases en archivos `.py`

```python
%%writefile manejador_csv.py
import csv

class ManejadorCSV:
    def __init__(self, ruta_archivo: str):
        self.ruta_archivo = ruta_archivo
        self.datos = []
        self.encabezados = []

    def importar(self):
        with open(self.ruta_archivo, mode='r', encoding='utf-8') as f:
            lector = csv.reader(f)
            self.encabezados = next(lector)
            self.datos = list(lector)
        print(f"CSV importado correctamente: {len(self.datos)} registros cargados.")

    def modificar_registro(self, id_registro: str, columna: str, nuevo_valor: str):
        # Buscar el índice de la columna solicitada
        idx_col = self.encabezados.index(columna)
        modificado = False
        for fila in self.datos:
            if fila[0] == id_registro:  # El Id está en la primera columna
                fila[idx_col] = nuevo_valor
                modificado = True
                break
        if modificado:
            print(f"Registro ID {id_registro} modificado: {columna} -> {nuevo_valor}.")
        else:
            print(f"No se encontró el registro con ID {id_registro}.")

    def guardar(self, nueva_ruta: str):
        with open(nueva_ruta, mode='w', newline='', encoding='utf-8') as f:
            escritor = csv.writer(f)
            escritor.writerow(self.encabezados)
            escritor.writerows(self.datos)
        print(f"Nuevo archivo CSV guardado exitosamente en: {nueva_ruta}")
```

```python
%%writefile manejador_json.py
import json

class ManejadorJSON:
    def __init__(self, ruta_archivo: str):
        self.ruta_archivo = ruta_archivo
        self.datos = []

    def importar(self):
        with open(self.ruta_archivo, mode='r', encoding='utf-8') as f:
            self.datos = json.load(f)
        print(f"JSON importado correctamente: {len(self.datos)} objetos cargados.")

    def agregar_registro(self, nuevo_registro: dict):
        self.datos.append(nuevo_registro)
        print(f"Nuevo registro agregado al JSON. Total de registros: {len(self.datos)}")

    def guardar(self, nueva_ruta: str):
        with open(nueva_ruta, mode='w', encoding='utf-8') as f:
            json.dump(self.datos, f, indent=2, ensure_ascii=False)
        print(f"Nuevo archivo JSON guardado exitosamente en: {nueva_ruta}")
```

```python
%%writefile generador_resumen_txt.py
class GeneradorResumenTXT:
    def __init__(self, url_csv: str, url_json: str,
                 archivo_csv_original: str, archivo_json_original: str,
                 registros_csv: int, registros_json: int,
                 cambios_csv: str, cambios_json: str,
                 ruta_salida_csv: str, ruta_salida_json: str,
                 clases_metodos: list):
        self.url_csv = url_csv
        self.url_json = url_json
        self.archivo_csv_original = archivo_csv_original
        self.archivo_json_original = archivo_json_original
        self.registros_csv = registros_csv
        self.registros_json = registros_json
        self.cambios_csv = cambios_csv
        self.cambios_json = cambios_json
        self.ruta_salida_csv = ruta_salida_csv
        self.ruta_salida_json = ruta_salida_json
        self.clases_metodos = clases_metodos

    def generar_reporte(self, ruta_resumen: str = "resumen.txt"):
        with open(ruta_resumen, "w", encoding="utf-8") as f:
            f.write("=" * 50 + "\n")
            f.write("        REPORTE DE PROCESAMIENTO AE2\n")
            f.write("=" * 50 + "\n\n")
            f.write("1. URL DE ORIGEN (Hugging Face):\n")
            f.write(f"   - CSV: {self.url_csv}\n")
            f.write(f"   - JSON: {self.url_json}\n\n")
            f.write("2. ARCHIVOS IMPORTADOS:\n")
            f.write(f"   - CSV: {self.archivo_csv_original}\n")
            f.write(f"   - JSON: {self.archivo_json_original}\n\n")
            f.write("3. REGISTROS PROCESADOS:\n")
            f.write(f"   - CSV: {self.registros_csv}\n")
            f.write(f"   - JSON: {self.registros_json}\n\n")
            f.write("4. MODIFICACIONES REALIZADAS:\n")
            f.write(f"   - CSV: {self.cambios_csv}\n")
            f.write(f"   - JSON: {self.cambios_json}\n\n")
            f.write("5. ARCHIVOS GUARDADOS:\n")
            f.write(f"   - CSV: {self.ruta_salida_csv}\n")
            f.write(f"   - JSON: {self.ruta_salida_json}\n\n")
            f.write("6. CLASES Y MÉTODOS UTILIZADOS:\n")
            for item in self.clases_metodos:
                f.write(f"   - {item}\n")
        print(f"Resumen generado correctamente en: '{ruta_resumen}'")
```

## 3. Procesamiento del CSV: importar, modificar y guardar

```python
from manejador_csv import ManejadorCSV

print("=== PROCESANDO CSV ===")
csv_handler = ManejadorCSV(archivo_csv_original)
csv_handler.importar()
print("Encabezados:", csv_handler.encabezados)

# Modificación: actualizar un valor
id_csv = "1"
columna_csv = "SepalLengthCm"
nuevo_valor_csv = "5.5"

fila = next(f for f in csv_handler.datos if f[0] == id_csv)
fila_antes = list(fila)
print("ANTES:  ", fila_antes)

csv_handler.modificar_registro(id_registro=id_csv, columna=columna_csv, nuevo_valor=nuevo_valor_csv)
print("DESPUÉS:", fila)

valor_anterior_csv = fila_antes[csv_handler.encabezados.index(columna_csv)]
cambios_csv = f"Registro Id {id_csv}: {columna_csv} cambió de {valor_anterior_csv} a {nuevo_valor_csv}"

csv_handler.guardar(archivo_csv_modificado)
```

## 4. Procesamiento del JSON: importar, modificar y guardar

```python
from manejador_json import ManejadorJSON

print("=== PROCESANDO JSON ===")
json_handler = ManejadorJSON(archivo_json_original)
json_handler.importar()
registros_json_antes = len(json_handler.datos)
print("Último registro ANTES:", json_handler.datos[-1])

# Modificación: agregar un registro
nuevo_item = {
    "instruction": "Write a Python function that returns the sum of two numbers.",
    "input": "",
    "output": "def sumar(a, b):\n    return a + b"
}
json_handler.agregar_registro(nuevo_item)
print("Último registro DESPUÉS:", json_handler.datos[-1])

cambios_json = f"Se agregó un registro nuevo. Registros: {registros_json_antes} -> {len(json_handler.datos)}"

json_handler.guardar(archivo_json_modificado)
```

## 5. Verificación de los archivos nuevos

```python
print("=== VERIFICACIÓN DE ARCHIVOS GUARDADOS ===")
verif_csv = ManejadorCSV(archivo_csv_modificado)
verif_csv.importar()
print("Fila modificada en el nuevo CSV:", next(f for f in verif_csv.datos if f[0] == id_csv))

verif_json = ManejadorJSON(archivo_json_modificado)
verif_json.importar()
print("Último registro en el nuevo JSON:", verif_json.datos[-1])
```

## 6. Generación y visualización del resumen

```python
from generador_resumen_txt import GeneradorResumenTXT

clases_metodos = [
    "ManejadorCSV: importar(), modificar_registro(), guardar()",
    "ManejadorJSON: importar(), agregar_registro(), guardar()",
    "GeneradorResumenTXT: generar_reporte()"
]

resumidor = GeneradorResumenTXT(
    url_csv=url_csv_origen,
    url_json=url_json_origen,
    archivo_csv_original=archivo_csv_original,
    archivo_json_original=archivo_json_original,
    registros_csv=len(csv_handler.datos),
    registros_json=len(json_handler.datos),
    cambios_csv=cambios_csv,
    cambios_json=cambios_json,
    ruta_salida_csv=archivo_csv_modificado,
    ruta_salida_json=archivo_json_modificado,
    clases_metodos=clases_metodos
)
resumidor.generar_reporte("resumen.txt")

with open("resumen.txt", "r", encoding="utf-8") as f:
    print(f.read())
```
